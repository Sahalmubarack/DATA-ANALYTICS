```python
import pandas as pd
```

```python
#creating a series from a list 
s = pd.Series([10,20,30,40,50])
print(s)
```

```python
# creating a series from a list of strings 

s = pd.Series(["apple","banana","mango","orange"])
print (s)
```

```python
#series with a custom index

s = pd.Series(
    [100,200,300,400],
    index=["a","b","c","d"]
)
print(s)
```

```python
#7. Creating a Series from a Dictionary


data = {
    "a":100,
    "b":200,
    "c":300,
    "d":400
}

s = pd.Series(data)
print(s)
```

```python
#8. Accessing Series Elements - iloc

s = pd.Series([10, 20, 30, 40, 50])
print(s.iloc[2])
```

```python
#9. loc - Label-Based Access

s=pd.Series(
    [100,200,300,400],
    index=["a","b","c","d"]
)
print(s.loc["c"])
```

```python
#10. Slicing a Series

s=pd.Series([10,20,30,40,50])
print(s.iloc[:3])
```

```python
#11. First and Last Elements

# First element
print(s.iloc[0])
# Last element
print(s.iloc[-1])
# First three elements
print(s.iloc[:3])
# Last two elements
print(s.iloc[-2:])
```

```python
#12. Series Arithmetic Operations

s=pd.Series([10,20,30,40,50])
print(s+10)
```

```python
#13. Mathematical Functions

marks=pd.Series([75,88,92,65,80])

print(marks.sum()) 
print(marks.mean()) 
print(marks.max())
print(marks.min()) 
```

```python
#14. Important Statistical Methods

s = pd.Series([10, 20, 30, 40, 50])
print(s.sum())
print(s.mean())
print(s.median())
print(s.max())
print(s.min())
```

# Data Selection Using loc and iloc

```python
#1. loc (Label-Based Indexing)

import pandas as pd
data= {
"Name": ["Asha","Rahul","Anu"],
"Age": [20,21,19],
"City": ["Kochi","Delhi","Chennai"]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.loc[1])
```

```python
print(df.loc[0:1, ["Name","Age"]])
```

```python
#2. iloc (Integer Position-Based Indexing)

print(df.iloc[1])
```

```python
print(df.iloc[0:2,0:2])
```

```python
#Difference in Slicing

print(df.loc[0:1])
```

```python
#Using iloc
print(df.iloc[0:1])
```

```python
#Selecting a Single Value

print(df.loc[2,"City"])
```

```python
print(df.iloc[2,2])
```
