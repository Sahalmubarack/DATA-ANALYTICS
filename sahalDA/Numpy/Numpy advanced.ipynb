## data types

```python
a=np.array([1,2,3], dtype=float)
print(a)
print(a.dtype)
```

```python
a=np.array([True,False,True])
print(a.dtype)
```

## itemsize and nbytes

```python
a=np.array([1,2,3], dtype=np.int32)
print(a.itemsize)
print(a.nbytes)
```

```python
a = np.array([1.5,2.5,3.5,4.5,],dtype=np.float64)
print(a.itemsize)
print(a.nbytes)
```

## array copy vs view

```python
a = np.array([10,20,30,40])
b = a.copy()
b[0] = 100
print(a)
print(b)
```

```python
a = np.array([10,20,30,40])
c = a.view()
c[0] = 100
print(a)
print(c)
```

```python
a = np.array([1,2,3,4])
b = a.copy()
c = a.view()
print(b.base)
print(c.base)
```

## array reshaping

```python
a = np.array([1, 2, 3, 4, 5, 6])
print(a.reshape(2, 3))
```

```python
a = np.array([[1, 2],
             [3, 4]])
print(a.reshape(4))
```

## flatten array

```python
a = np.array([[1, 2, 3],
             [4, 5, 6]])
print(a.flatten())
```

```python
a = np.array([[1, 2],
              [3, 4]])
b = a.flatten()
b[0] = 100
print(a)
print(b)
```

```python
a = np.array([[1, 2],
              [3, 4]])
b = a.flatten()
c = a.ravel()
b[0] = 100
print(a)
print(b)
print(c)
```
