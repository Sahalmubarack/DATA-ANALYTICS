## IDENTITY OPERATORS

```python
#1 check whether student1=["anu",20] and student2=student1 refer to the same object.

student1 = ["anu", 20]
student2 = student1
print(student1 is student2)
```

```python
#2 check whether list=[10,20,30] and list2=[10,20,30] are the same object.

list1 = [10,20,30]
list2 = [10,20,30]
print(list1 is list2)
```

```python
#3 check whether employee1 = {"name" : "rahul"} and employee2 = employee1 refer to the same object.

employee1 = {"name" : "rahul"}
employee2 = employee1
print(employee1 is employee2)
```

```python
#4 Check whether marks1 = [85, 90, 78] and marks2 = [85, 90, 78] are different objects.

mark1 = [85, 90, 78]
mark2 = [85, 90, 78]
print(mark1 is not mark2)
```

```python
#5 Check whether result = None refers to the None object.

result=None
print(result is None)
```
