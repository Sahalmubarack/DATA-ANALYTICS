```python
#1. Check whether a number is positive, negative, or zero.

x=int(input("enter a number"))
if x>0 :
    print("number is positive")
elif x<0 :
    print("number is negative")
else :
    print("number is zero")
```

```python
#2. Check whether a number is even or odd.

x=int(input("enter a number"))
if x %2==0:
    print("number is even")
else :
    print("number is odd")
```

```python
#6. Check whether a year is a leap year.

year=int(input("enter the year"))
if year %400==0:
    print("this year is leap year")
elif year %100==0:
    print ("this year is not leap year")
elif year %4==0:
    print ("this year is leap year")
else :
    print ("this year is not a leap year")
```

```python
#11. Check whether a character is an uppercase letter, lowercase letter, or digit.

x=input("enter a character")
if x.isupper() :
    print ("this character is an uppercase letter")
elif x.islower() :
    print ("this character is a lowercase letter")
elif x.isdigit() :
    print ("this character is a digit")
else :
    print ("invalid")
```

```python
#14. Create a simple calculator using the user's choice of operator (+, -, *, /).

a=int(input("enter 1st number"))
b=int(input("enter 2nd number"))
operator=input("enter an operator (+,-,*,/) :")
if operator=="+":
    print ("result =",a+b)
elif operator=="-":
    print ("result =",a-b)
elif operator=="*":
    print ("result =",a*b)
elif operator=="/":
    if b!=0:
        print ("result",a/b)
    else :
        print ("divition not possible")
else :
    print ("invalid operator")
        
```
