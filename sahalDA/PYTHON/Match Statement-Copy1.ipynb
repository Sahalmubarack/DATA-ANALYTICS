```python
#1. Display the full form of a department code using match ( CS → Computer Science, IT → Information Technology, EC → Electronics, ME → Mechanical).

department=input("enter department code").lower()
match department :
    case "cs" :
        print("computer science")
    case "it" :
        print("information technology")
    case "ec" :
        print("electronnics")
    case "me" :
        print("mechanical")
    case _:
        print("invalid department")
```

```python
#2. Display the month name based on a month number (1–12).

month=int(input("enter a month number"))
match month :
    case 1:
        print("january")
    case 2:
        print("february")
    case 3:
        print("march")
    case 4:
        print("april")
    case 5:
        print("may")
    case 6:
        print("june")
    case 7:
        print("july")
    case 8:
        print("august")
    case 9:
        print("september")
    case 10:
        print("october")
    case 11:
        print("november")
    case 12:
        print("december")
    case _:
        print("invalid month number")
```

```python
#3. Create a simple calculator using match ( + , , , / ).

a=int(input("enter 1st number"))
b=int(input("enter 2nd number"))
operator=input("enter a operator (+,-,*,/):")
match operator :
    case "+":
        print ("result", a+b)
    case "-":
        print ("result", a-b)
    case "*":
        print ("result", a*b)
    case "/":
        if b!=0:
            print ("result", a/b)
        else :
            print ("divition not possible")
    case _ :
        print ("invalid operator")
```
