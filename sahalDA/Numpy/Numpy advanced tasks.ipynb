```python
#2. Given the array: a = np.array([1, 2, 3, 4], dtype=np.int32) Find and print the data type, item size, and total memory used by the array.

a = np.array([1, 2, 3, 4], dtype=np.int32)
print(a.dtype)
print(a.itemsize)
print(a.nbytes)
```

```python
#4. Given the array:a = np.array([[1, 2], [3, 4]]) Create a view of the array, change the first element of the view to 100 , and print both the original array and the view. Check the base attribute.

a = np.array([[1, 2], [3, 4]])
b= a.view()
b[0][0]=100
print(a)
print(b)
print(b.base)
```

```python
# 17.Given the array:
[[2, 4, 6],
[8, 10, 12]]
#Find the sum of the values in each row and the sum of the values in each column.

a=np.array([[2,4,6],
            [8,10,12]])
print("column wise")           
print(np.sum(a,axis=0))
print("row wise")
print(np.sum(a,axis=1))           
```
